```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score

df = pd.read_csv('data.csv')

X = df[['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model year', 'origin']]
# y = X['origin']
# X.drop('origin', axis=1, inplace=True)
# Y = df["origin"]
y = X['origin']
X.drop('origin', axis=1, inplace=True)

X_dummies = pd.get_dummies(X, columns=['model year'], dtype=float)

# Splitting data in train and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X_dummies, y, test_size=0.30, random_state=1
)

# Define the parameters of the tree to iterate over
max_depth_values = np.arange(2, 11, 3)
max_leaf_nodes_values = np.arange(10, 21, 10)


# Initialize variables to store the best model and its performance
diff_f1_score = float('inf')
best_max_depth = None
best_max_leaf_node = None

# Iterate over all combinations of the specified parameter values
for max_depth in max_depth_values:
    for max_leaf_nodes in max_leaf_nodes_values:
        
        # Initialize the tree with the current set of parameters
        dtree = DecisionTreeClassifier(
            max_depth=max_depth,
            max_leaf_nodes=max_leaf_nodes,
            #min_samples_split=min_samples_split,
            #class_weight='balanced',
            random_state=1
        )

        # Fit the model to the training data
        dtree.fit(X_train, y_train)

        # Make predictions on the training and test sets
        y_train_pred = dtree.predict(X_train)
        y_test_pred = dtree.predict(X_test)

        # Calc F1 score
        train_f1 = f1_score(y_train, y_train_pred, average='macro')
        test_f1 = f1_score(y_test, y_test_pred, average='macro')

        if train_f1 > test_f1:
            curr_diff = abs(train_f1 - test_f1)
            if curr_diff < diff_f1_score:
                diff_f1_score = curr_diff
                best_max_depth = max_depth
                best_max_leaf_node = max_leaf_nodes

```
